```python
%sql
select order_id,
books,
filter(books, x -> x.quantity >= 2) as multiple_copy
from orders;

select order_id,multiple_copies from (select order_id,
filter(books, x -> x.quantity >= 2) as multiple_copies
from orders)
where size(multiple_copies) > 0;
```

```python
%sql
select order_id,
books,
transform(books, x -> x.subtotal * 1.2) as subtotals
from orders;
```

```python
%sql
create or replace function get_url(email string)
returns string
return concat("https://www.",split(email,"@")[1])
```

```python
%sql
select email,get_url(email) from customers;
```
